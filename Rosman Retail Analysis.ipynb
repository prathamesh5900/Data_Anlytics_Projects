### Log transformation 

```python
data["Sales_log"] = np.log1p(data["Sales"])
```

```python
plt.figure(figsize=(10,8))
sns.histplot(data["Sales_log"] ,bins=50 , kde=True)
plt.title("Log-transformed sales distribution")
plt.show()
```

The distribution has changed from right-skewed to approximately normal (bell-shaped).

Most values are now centered around 8 to 9, with the tail ends symmetrically tapering off.

This transformation makes the data more suitable for linear models, which assume normally distributed residuals.



```python
sns.boxplot(data["Sales_log"], orient="h")
plt.show()
```

The boxplot shows:

A tight interquartile range (IQR), indicating reduced spread.

Fewer extreme outliers compared to the original scale.

Though some outliers still exist, their influence on the model is now reduced due to the compressed scale.

**Log transformation reduces the effect of outliers,making model more stable**

## Misssing values

```python
data.info()
```

```python
missing_value = data.isnull().sum()
```

```python
missing_value[missing_value>0]
```

There are only 6 columns which are related to competition and promotion have missing values 

**Since the missing values in these columns would be store specific , need to check whether missing entries are consistent per store**

```python
cols_to_check = [
    'CompetitionDistance',
    'CompetitionOpenSinceMonth',
    'CompetitionOpenSinceYear',
    'Promo2SinceWeek',
    'Promo2SinceYear',
    'PromoInterval'
]


for col in cols_to_check:
    print(f"COlumn:{col}")
    unique_values_per_store = data.groupby('Store')[col].nunique(dropna=False)
    print(f"Stores with consistent value (or missing) for '{col}': {(unique_values_per_store == 1).sum()} / {data['Store'].nunique()}")

          
```

All stores show nunique == 1, it means that column is consistent within each store

1.Missing CompetitionDistance ⇒ doesn’t mean "no competition nearby"; it means "distance data is missing", so we fill it with median distance to avoid distorting the model.

2.Missing CompetitionOpenSinceMonth/Year ⇒ assume no competition start date, so fill with 0 (logical default).

```python
data["CompetitionDistance"] = data["CompetitionDistance"].fillna(data["CompetitionDistance"].median())


data["CompetitionOpenSinceMonth"] = data["CompetitionOpenSinceMonth"].fillna(0)
data["CompetitionOpenSinceYear"] = data["CompetitionOpenSinceYear"].fillna(0)
```

**We will check whether promo2 column have any influence in Promo related columns**

Checking if all promo2 related missing values are for Promo2 == 0

```python
print("Promo2SinceWeek:")
print(data[data['Promo2SinceWeek'].isna()]['Promo2'].value_counts())

print("\nPromo2SinceYear:")
print(data[data['Promo2SinceYear'].isna()]['Promo2'].value_counts())

print("\nPromoInterval:")
print(data[data['PromoInterval'].isna()]['Promo2'].value_counts())
```

**From here we can infer that all missing values were not participated in promo2 , we can replace them with zero or none.**

```python
data['Promo2SinceWeek'] = data['Promo2SinceWeek'].fillna(0)
data['Promo2SinceYear'] = data['Promo2SinceYear'].fillna(0)
data['PromoInterval'] = data['PromoInterval'].fillna('None')
```

```python
data.isnull().sum()
```

**All missing values are filled now**

## Monthly and weekly sales analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd


data['Date'] = pd.to_datetime(data['Date'])

# Extract Month-Year for grouping
data['Month_Year'] = data['Date'].dt.to_period('M').astype(str)

# Aggregate monthly sales
monthly_sales = data.groupby('Month_Year')['Sales'].sum().reset_index()

# Day of Week mapping
day_labels = {
    1: 'Sunday', 2: 'Monday', 3: 'Tuesday', 4: 'Wednesday',
    5: 'Thursday', 6: 'Friday', 7: 'Saturday'
}

# Group by DayOfWeek
daily_sales = data.groupby('DayOfWeek')['Sales'].mean().reset_index()
daily_sales['DayName'] = daily_sales['DayOfWeek'].map(day_labels)
daily_sales = daily_sales.sort_values('DayOfWeek')

# Create subplots
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# 1. Line plot for monthly sales
sns.lineplot(data=monthly_sales, x='Month_Year', y='Sales', ax=axes[0], marker='o')
axes[0].set_title('Monthly Sales Trend')
axes[0].set_xlabel('Month-Year')
axes[0].set_ylabel('Total Sales')
axes[0].tick_params(axis='x', rotation=45)

# 2. Bar plot for sales by day of week
sns.barplot(x='DayName', y='Sales', data=daily_sales, palette='pastel', ax=axes[1])
axes[1].set_title("Average Sales by Day of Week")
axes[1].set_xlabel("Day of Week")
axes[1].set_ylabel("Average Sales")
axes[1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

1.November 2014 shows a major sales spike, indicating seasonal peaks—likely due to holidays or promotions.

2.Sales dipped significantly around mid-2015 (July), suggesting a seasonal or external slowdown.

3.Saturday and Sunday have the highest average sales, making weekends the most profitable days.

4.Friday has surprisingly the lowest average sales, highlighting a potential opportunity for improvement.
